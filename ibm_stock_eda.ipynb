```python
q1= df["4. close"].quantile(0.25)
q3= df["4. close"].quantile(0.75)
iqr= q3-q1
print(q1,q3,iqr)
lower= q1-1.5*iqr
upper= q3+1.5*iqr
sns.boxplot(df[["1. open","4. close"]])
result = [df[(df["4. close"] < lower) | (df["4. close"] > upper)]]
print(len(result[0]))
```
